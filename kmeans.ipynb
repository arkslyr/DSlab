```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
```

```python
data,_=make_blobs(n_samples=300)
print(data)
```

```python
kmeans=KMeans(n_clusters=5,random_state=42,n_init='auto').fit(data)
```

```python
labels=kmeans.labels_

#it contains the clusters assigned each datapoint specifying which cluster each datapoint belongs to
```

```python
set(labels)
```

```python
centroids=kmeans.cluster_centers_
#it stores the coordinates of the centroids for these five clusters
centroids
```

```python
plt.scatter(data[:,0],data[:,1],c=labels,cmap='viridis')
plt.scatter(centroids[:,0],centroids[:,1],color='red',marker='x')
plt.legend(list(labels))
plt.show()
#this will plot the data point and coloring each point according to cluster labels
```

```python
plt.scatter([2,6],[-4,6],color='cyan')
plt.show()
```
